```r
# Criando a matriz 2x2 do slide (Slide 13)
m <- matrix(1:4, nrow = 2)
rownames(m) <- c("lin1", "lin2")
colnames(m) <- c("c1", "c2")

# Vetor com 3 elementos
peso3 <- c(2, 3, 5)

# Operação com reciclagem
m * peso3

#Ao multiplicar a matriz 2x2(4 elementos) pelo vetor peso de 3 elementos, o R tentará realizar a reciclagem.
#Como o tamanho da matriz não é um múltiplo do tamanho do vetor, o cálculo será executado elemento a elemento por colunas, mas o R emitirá um warning.
#A reciclagem só ocorre de forma "silenciosa" quando o tamanho do objeto maior é múltiplo do menor (ex: 4 e 2).
#Como 4 não é múltiplo de 3, o R concluiu a última multiplicação reutilizando o primeiro item do vetor, mas emitiu o alerta longer object length is not a multiple of shorter object length para avisar que a reciclagem foi interrompida no meio de um ciclo.
```

```r
# 1. Criação do vetor nomeado com 3 frases
frases <- c(
  f1 = "recuperacao de informacao e ciencia de dados",
  f2 = "ciencia de dados utiliza processamento de texto",
  f3 = "recuperacao de texto na ciencia de dados"
)

# 2. Tokenização utilizando lapply e quebra por espaços (\s+)
tokens_lista <- lapply(frases, function(texto) {
  unlist(strsplit(texto, "\\s+"))
})

# 3. Transformação em vetor simples (unlist) e contagem dos termos
todos_tokens <- unlist(tokens_lista)
contagem <- table(todos_tokens)

# 4. Ordenação dos termos do maior para o menor
contagem_ordenada <- sort(contagem, decreasing = TRUE)

# Exibição do resultado
print(contagem_ordenada)
```

# **5 Missões**

```r
manchetes <- c(
"Porto de Santos bate recorde em julho - A Tribuna",
"cubatao registra melhora na qualidade do ar - A Tribuna",
"Guaruja tera nova linha de onibus em 2026 - A Tribuna",
"Sao Vicente inaugura escola no Parque Bitaru - A Tribuna",
"Santos e Guaruja discutem travessia de balsa - A Tribuna"
)

#1- Faxina:
print("Manchetes sem 'a tribuna'")
frases_limpas <- sub(" - A Tribuna$", "", manchetes)
frases_limpas

#2- Espaços:
print("Mudando dois ou mais espaços por um")
gsub("\\s{2,}", " ", frases_limpas)

#3- Ano:
print("Quais manchetes citam um ano (4 digitos)")
grep("[0-9]{4}", frases_limpas, value= TRUE)

#4- Municípios:
print("Quais manchetes citam Guarujá ou Cubatão, ignorando maiúsculas")
grep("Guaruja|Cubatao", frases_limpas, ignore.case = TRUE, value = TRUE)

#5- A pegadinha:
print("Com ''grepl('Santos', manchetes)'' acha 2 manchetes — mas uma
fala do Porto, não da cidade. Como você distinguiria? É possível só com regex?")
grepl("Santos", frases_limpas)
print("Seria possível remover a palavra porto da pesquisa com regex, porém com uma quantidade maior
de manchetes poderia ocorrer um erro... A regex reconhece apenas padrões de texto, não o significado
das palavras, então não distingui a utilização da palavra 'Santos'")
```
